This project analyses the financial performance of private start-up companies that went public 5 years ago. Financial information, such as income statements, balance sheets, and cash flows, is used to create the final data frame. Historical data is used to calculate the average closing price and total stock volume sold between 2020-2025.
Python is used as an ETL tool to extract the data from Yahoo Financials, perform data transformations to consolidate the necessary columns and load the data into an Excel sheet.




# 1. Import Libraries

```python
##Loading the libraries
import pandas as pd
import numpy as np
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import yfinance as yf
```

```python
pd.options.display.float_format = '{:.2f}'.format
```

# 2. Extract Initial Public Offering (IPO) Data

```python
## extracting IPO data for funding circle
ticker1 = yf.Ticker('FCH.L')
income_stmt_fch = ticker1.financials.T
bal_sheet_fch = ticker1.balance_sheet.T
cash_flow_fch = ticker1.cash_flow.T
```

```python
## extracting IPO data for One Saivngs Bank
ticker2 = yf.Ticker('LINV.L')
income_stmt_linv = ticker2.financials.T
bal_sheet_linv = ticker2.balance_sheet.T
cash_flow_linv = ticker2.cash_flow.T
```

```python
cash_flow_fch.head()
```

# 3. Get Finance & Revenue Data

## Income Statement Columns
1. Net Income
2. Operating Income & Expense : Operating Efficiency Ratio
3. Gross Profit: Gross Profit Margin
4. Cost of Revenue : Cost of Goods & Services
5. Revenue: Total Revenue, Revenue Growth Rate, Operating Efficiency Ratio, Profit Margin, Gross Profit Margin

```python
income_stmt_linv.columns
```

```python
#extract only columns that have a direct meaning or can be indirectly calculated without too many calculations
#Cost of Revenue : Cost of Goods and Services (CoGS)
fin_cols=['Net Income','Operating Income', 'Operating Expense','Gross Profit', 'Cost Of Revenue', 'Total Revenue']
income_fch = income_stmt_fch[fin_cols].copy()
income_linv =  income_stmt_linv[fin_cols].copy()
income_linv
```

```python
bal_sheet_fch.columns
```

## Balance Sheet Columns
1. Current Assest & liabilities : To show Current Ratio
2. Cash Cash Equivalent and Short Term Investments is considered as Cash Availability



```python
# Cash and Cash Equivalents and Short Term Investments is considered as the "Cash available". cash equivalents + short-term investments,
# such as marketable securities (like short-term bonds or CDs) that can be quickly converted to cash

#Initially thought of using Cash and Cash Equivalents which is "pure cash " as "cash Available" but then chose a broader term like the above term"
bal_cols =['Current Assets','Current Liabilities','Cash Cash Equivalents And Short Term Investments']
bal_fch = bal_sheet_fch[bal_cols].copy()
bal_linv = bal_sheet_linv[bal_cols].copy()
bal_fch.rename(columns={'Cash Cash Equivalents And Short Term Investments':'Cash Available'}, inplace=True)
bal_linv.rename(columns={'Cash Cash Equivalents And Short Term Investments':'Cash Available'}, inplace=True)
```

```python
bal_fch
```

```python
cash_flow_fch.columns
```

## Cash Flow Columns
1. Operating Cash Flow: Cash inflow - Cash Outflow
2. Capital Expenditure: Total money spent on company growth and operation for future

```python
# operating cash flow is considered as the cash flow (cash inflow - cash outflow) essential to calculate the burn rate and the runaway
cash_fch  = cash_flow_fch[['Operating Cash Flow','Capital Expenditure']].copy()
cash_linv  = cash_flow_linv[['Operating Cash Flow','Capital Expenditure']].copy()
```

# 4. Combining the data columns selected from financials into one dataframe

```python
#combining the selected columns
fch = income_fch.join([bal_fch, cash_fch], how='outer')
fch.fillna(0, inplace=True)
linv = income_linv.join([bal_linv, cash_linv], how='outer')
linv.fillna(0, inplace=True)
```

```python
fch.index.year
```

```python
# Creating a year column from the date
fch['Year'] = fch.index.year
linv['Year'] = linv.index.year
```

# 5. Fetch the stock data to generate the average close value and total volume

```python
stock_fch = ticker1.history(start = '2020-01-01', end ='2024-12-31')
stock_linv = ticker2.history(start = '2020-01-01', end ='2024-12-31')
```

```python
#extracting the avg closing value and the total volume of stocks sold for the year
stock_value_fch = stock_fch.resample('YE').agg({'Close':'mean',
                          'Volume':'sum'}).rename(columns={'Close':'Avg Close Price',
                                                           'Volume':'Total Volume'})
stock_value_linv = stock_linv.resample('YE').agg({'Close':'mean',
                          'Volume':'sum'}).rename(columns={'Close':'Avg Close Price',
                                                           'Volume':'Total Volume'})
```

```python
stock_value_fch.index.year
```

```python
#creating the year column for stocks
stock_value_fch['Year'] = stock_value_fch.index.year
stock_value_linv['Year'] = stock_value_linv.index.year
```

# 6. Merge the financials and the stock data into a final document

```python
# combining the financials and the stock data on Year using merge statement
fch_df = pd.merge(fch, stock_value_fch, on='Year').set_index('Year')
linv_df = pd.merge(linv, stock_value_linv, on='Year').set_index('Year')
```

# 7. Load the dataframe to a csv file

```python
fch_df.drop([2020,2021], axis=0, inplace=True)
fch_df
```

```python
linv_df.drop([2021], inplace=True)
linv_df
```

```python
fch_df.columns = fch_df.columns.str.replace(' ', '_')
linv_df.columns = linv_df.columns.str.replace(' ', '_')
```

```python
fch_df.to_csv('fch.csv', index=True)
print("FCH Data successfully exported to Excel")

linv_df.to_csv('linv.csv', index=True)
print("LINV Data successfully exported to Excel")
```
